```python
from pathlib import Path
import geopandas as gpd

# Locate project root automatically
cwd = Path.cwd()
root = cwd if cwd.name == "damage-control-thesis" else next(cwd.rglob("damage-control-thesis"))

# Find the extracted .shp file
shp_path = next((root / "data/external/unosat").glob("*.shp"))

# Load vector data and reproject to match Sentinel-2 (UTM 37N)
gdf = gpd.read_file(shp_path)
gdf_utm = gdf.to_crs("EPSG:32637")

print("Original CRS:", gdf.crs)
print("Reprojected CRS:", gdf_utm.crs)
print("Total damaged structure records:", len(gdf_utm))
print("Columns:", list(gdf_utm.columns))
gdf_utm.head(3)
```
