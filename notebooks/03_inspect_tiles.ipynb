```python
import rasterio

with (
    rasterio.open("data/raw/pre/pre_r00_c00.tif") as pre,  
     rasterio.open("data/raw/post/post_r00_c00.tif") as post
):
    for name, ds in (("pre",pre), ("post",post)):
        print(name, ds.crs, ds.width, ds.height, ds.count, ds.dtypes[0])
        print(ds.transform)
    print("same CRS:", pre.crs == post.crs)
    print("same size:", (pre.width, pre.height) == (post.width, post.height))
    print("same transform:",pre.transform == post.transform)
```
